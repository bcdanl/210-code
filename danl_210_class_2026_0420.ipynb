```python
dict_data = {"Name": ["Donna", "Donna", "Mike", "Mike"],
             "Department": ["ECON", "DANL", "ECON", "DANL"],
             "2022": [1, 2, 3, 1],
             "2023": [2, 3, 4, 2],
             "2024": [5, 1, 2, 2]}
df = pd.DataFrame(dict_data)

df
```

```python
df_longer = df.melt(id_vars=["Name", "Department"],
                    var_name="Year",
                    value_name="Number of Courses")

df_longer
```

```python
df_wider = (
    df_longer
    .pivot(
        index = ["Name", "Department"],
        columns = 'Year',
        values = 'Number of Courses'
    )
    .reset_index()
)

df_wider
```

# Question 1

Reshape ny_pincp to a longer format.

```python
ny_pincp = pd.read_csv('https://bcdanl.github.io/data/NY_pinc_wide.csv')
ny_pincp
```

```python
ny_pincp_longer = (
    ny_pincp
    .melt(
        id_vars = ['fips', 'geoname'],
        var_name= 'Year',
        value_name= 'Income'
    )
)

ny_pincp_longer
```

```python
ny_pincp_longer = (
    ny_pincp
    .query('geoname != "New York"')
    .melt(
        id_vars = ['fips', 'geoname'],
        var_name= 'Year',
        value_name= 'Income'
    )
)

ny_pincp_longer
```
