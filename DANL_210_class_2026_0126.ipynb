# Dictionary

```python
city_to_temp = {
    "Paris": 28,
    "London": 22,
    "New York": 18,
    "Seoul": 29,
    "Rochester": 10
}
city_to_temp
```

```python
type(city_to_temp)
```

```python
city_to_temp["Paris"]          # 🔍 look up a value by key
```

```python
city_to_temp["London"] = 32    # ✏️ update a value
city_to_temp
```

```python
# keys(), values(), items()
city_to_temp.keys()        # 🔑 all keys
```

```python
type(city_to_temp.keys())
```

```python

city_to_temp.values()      # 🌡️ all values
```

```python

city_to_temp.items()       # 🧾 (key, value) pairs
```
