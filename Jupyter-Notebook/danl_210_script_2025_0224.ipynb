```python
df_wide = pd.DataFrame({
    'Weekday': ['Tuesday', 'Wednesday'],
    'Miami': [80, 83],
    'Rochester': [57, 62],
    'St. Louis': [71, 75]
})

df_long = pd.DataFrame({
    'Weekday': ['Tuesday', 'Wednesday', 'Tuesday', 'Wednesday', 'Tuesday', 'Wednesday'],
    'City': ['Miami', 'Miami', 'Rochester', 'Rochester', 'St. Louis', 'St. Louis'],
    'Temperature': [80, 83, 57, 62, 71, 75]
})
```

```python
df_long
```

```python
df_wide['diff_mia_roc'] = df_wide['Miami'] - df_wide['Rochester']
df_wide
```

```python
df_wide_to_long = (
    df_wide
    .melt()
)

df_wide_to_long
```

```python
df_wide
```

```python
df_wide_to_long = (
    df_wide
    .melt(id_vars = 'Weekday')
)

df_wide_to_long
```

```python
df_wide_to_long = (
    df_wide
    .melt(id_vars = 'Weekday',
          var_name= 'City',
          value_name= 'Temperature')
)

df_wide_to_long
```

```python
df_long
```

```python
df_long_to_wide = (
    df_long
    .pivot(
        index='Weekday',
        columns='City',
        values='Temperature'
    )
    .reset_index()
)

df_long_to_wide
```

```python
dict_data = {"Name": ["Donna", "Donna", "Mike", "Mike"],
             "Department": ["ECON", "DANL", "ECON", "DANL"],
             "2018": [1, 2, 3, 1],
             "2019": [2, 3, 4, 2],
             "2020": [5, 1, 2, 2]}
df = pd.DataFrame(dict_data)
df
```

```python
df_longer = df.melt(id_vars=["Name", "Department"],
                    var_name="Year",
                    value_name="Number of Courses")
df_longer
```

```python
(
    df_longer
    .pivot(
        index = ['Name', 'Department'],
        columns= 'Year',
        values= 'Number of Courses'
    )
    .reset_index()
)
```
