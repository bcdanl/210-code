```python
(
    fortune
    .groupby('Sector')
    .agg(
        Revenue_IQR = ('Revenues_M', lambda s: s.quantile(.75) - s.quantile(.25) ),
        Profit_IQR = ('Profits_M', lambda s: s.quantile(.75) - s.quantile(.25) ),
        Number_of_Employees_IQR = ('Number_of_Employees', lambda s: s.quantile(.75) - s.quantile(.25) )
    )
)
```

```python
(
    fortune
    .groupby('Sector')
    .agg(
        Revenue_Q1 = ('Revenues_M', 'quantile(.25)')
    )
)
```

```python
def IQR(s):
  return s.quantile(.75) - s.quantile(.25)
```

```python
(
    fortune
    .groupby('Sector')
    .agg(
        Revenue_IQR = ('Revenues_M', IQR)
    )
)
```

```python
fortune['Sector_rev_mean'] = (
    fortune.groupby('Sector')['Revenues_M'].transform('mean')
)
```
