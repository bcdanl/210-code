```python
emp = pd.read_csv(
    "https://bcdanl.github.io/data/employment.csv"
)

emp
```

```python
emp.info()
```

```python
emp = emp.astype(
    {
        'Gender':'category',
        'Start Date': 'datetime64[ns]',
        'Mgmt': bool,
        'Team': 'category'
    }
)
```

```python
emp.info()
```

## Q1
Write code to count how many employees belong to each team.



```python
q1 = emp['Team'].value_counts()

q1
```

## Q2
Write code to display the proportion of employees in each gender category.



```python
q2 = emp['Gender'].value_counts(normalize = True, dropna = False)
q2
```

## Q3
Write code to identify the team with the largest portion of senior managers.



```python
q3 = (
    emp[['Team' ,'Mgmt']]
    .value_counts(normalize = True)
    .reset_index()
    .query('Mgmt == True')
    .nlargest(1, 'proportion')
)

q3
```
