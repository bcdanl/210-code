```python
beer_mkt = pd.read_csv('https://bcdanl.github.io/data/beer_markets.csv')
beer_mkt
```

## Question 1
Provide (1) seaborn code and (2) a simple comment to describe how the distribution of price_per_floz varies by brand.

```python
sns.boxplot(data = beer_mkt,
            x = 'price_per_floz',
            y = 'brand',
            hue = 'brand')
```

```python

(
 sns.FacetGrid(
       data = beer_mkt,
       row='brand')
 .map(sns.histplot, 'price_per_floz')
 )
```

```python
plt.figure(figsize=(10, 6))  # width=10 inches, height=6 inches

(
 sns.FacetGrid(
       data = beer_mkt,
       col='brand')
 .map(sns.histplot, 'price_per_floz')
 )

plt.show()
```

```python
# Setting the seaborn's figure size
sns.set(rc={'figure.figsize': (10, 6)})
```
