```python
lst_students = [
    {"name": "Ava",  "major": "DANL", "score": 91},
    {"name": "Ben",  "major": "ECON", "score": 84},
    {"name": "Cara", "major": "DANL", "score": 76},
]

lst_students
```

Use a list comprehension to create danl_names, the names of students whose major is "DANL".

```python
danl_names = [student['name'] for student in lst_students if student['major'] == 'DANL' ]
danl_names
```

Use a dictionary comprehension to create danl_scores mapping name → score for only "DANL" students.

```python
danl_scores = { student['name'] : student['score'] for student in lst_students if student['major'] == 'DANL'}
```

```python
danl_scores
```
