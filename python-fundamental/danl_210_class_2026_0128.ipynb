# `if` statements

```python
name = "Geneseo"
score = 99

if name != "Geneseo" and score > 90:
  print("You are not called Geneseo and you have a high score")
```

```python

if name == "Geneseo" and score > 90:
    print("Geneseo, you achieved a high score.")
```
