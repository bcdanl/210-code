# `try`-`except`

```python
short_list = [1, 2, 3]
positions = [0, 1, 5, 2]   # 5 is out of range

for i in positions:
    print(short_list[i])
```

```python
short_list = [1, 2, 3]
positions = [0, 1, 5, 2]   # 5 is out of range

for i in positions:
    try:
        print(short_list[i])
    except:
        print("Index error:", i, "is not between 0 and", len(short_list) - 1)
```
