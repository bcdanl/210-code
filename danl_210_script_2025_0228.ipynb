# Join - Duplicate Keys

```python
import pandas as pd
```

```python
x = pd.DataFrame({
    'key':[1, 2, 2, 3],
    'val_x':['x1', 'x2', 'x3', 'x4']})

y = pd.DataFrame({
    'key':[1, 2],
    'val_y':['y1', 'y2'] })
one_to_many = x.merge(y, on='key',
                         how='left')
```

```python
x
```

```python
y
```

```python
x.merge(y, on = 'key', how = 'left')
```

```python
x = pd.DataFrame({
  'key':[1, 2, 2, 3],
  'val_x':['x1','x2','x3','x4']})

y = pd.DataFrame({
  'key': [1, 2, 2, 3],
  'val_y': ['y1', 'y2', 'y3', 'y4'] })
```

```python
x
```

```python
y
```

```python
x.merge(y, on = 'key', how = 'left')
```

```python
x = pd.DataFrame({
  'key_x': [1, 2, 3],
  'val_x': ['x1', 'x2', 'x3']
})


y = pd.DataFrame({
  'key_y': [1, 2],
  'val_y': ['y1', 'y2'] })
```

```python
x
```

```python
y
```

```python
keys_xy = x.merge(y,
                  left_on = 'key_x',
                  right_on = 'key_y',
                  how = 'left')
```

```python
keys_xy
```
